```python
import pandas as pd
```

```python
df = pd.read_csv('News.csv', engine = 'python', on_bad_lines= 'skip')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## get the independent features
x = df.drop('class' ,axis = 1)
```

```python
y = df['class']
```

```python
## check whather dataset is balanced or not
y.value_counts()
```

```python
import tensorflow as tf
```

```python
tf.__version__
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.layers import LSTM
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Bidirectional
```

```python
#vocabulary size
voc_size = 5000
```

```python
messages = x.copy()
```

```python
messages
```

```python
import nltk
```

```python
import re
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
corpus =[]
for i in range(0 ,len(messages)):
  review = re.sub('[^a-zA-Z]',' ',messages['title'][i])
  review = review.lower()
  review = review.split()
  review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)
```

```python
corpus
```

```python

# apply one hot encoding
onehot_repr = [one_hot(words,voc_size) for words in corpus]
onehot_repr
```

```python
#  then pass to Embedding Representation
sent_length = 20

embedded_docs=pad_sequences(onehot_repr, padding = 'pre', maxlen = sent_length)
embedded_docs
```

```python
# create model
Embedding_vector_features = 40
model = Sequential()
model.add(Embedding(voc_size, Embedding_vector_features, input_shape=(sent_length,)))
model.add(Bidirectional(LSTM(100)))
model.add(Dense(1, activation = 'sigmoid'))
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
print(model.summary())
```

```python
import numpy as np
x_final = np.array(embedded_docs)
y_final = np.array(y)
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train , x_test , y_train ,y_test = train_test_split(x_final , y_final , test_size = 0.33,random_state = 42)
```

```python
#model trainning
model.fit(x_train ,y_train , validation_data = (x_test , y_test),epochs = 10 ,batch_size = 32)
```

```python
y_pred = model.predict(x_test)
```

```python
import numpy as np
y_pred = np.where(y_pred>=0.5 , 1,0)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
confusion_matrix(y_test , y_pred)
```

```python
print(accuracy_score(y_test , y_pred))
```
